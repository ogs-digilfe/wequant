```python
# set para
# シミュレーション結果ファイル名のみを指定
sim_file = "threshold-20261025172241.csv"
```

```python
# set path
from pathlib import Path
import os
FP = Path(os.getcwd()).parent.parent / "data" / "output" / "sim" / sim_file

# set objects
import polars as pl

df = pl.read_csv(
    FP,
    try_parse_dates=True,
    schema_overrides={"code": pl.String},
)
```

```python
s_data_num = f"データ数: {df["updown"].count()}"
s_mean = f"平均: {df["updown"].mean():.2f}%, {df["benchmark"].mean():.2f}%"
s_win_rate = f"勝率: {df.filter(pl.col("updown")>0)["updown"].count() / df["updown"].count() * 100:.2f}%, {df.filter(pl.col("benchmark")>0)["benchmark"].count() / df["benchmark"].count() * 100:.2f}%"

print(s_data_num)
print(s_mean)
print(s_win_rate)
```
