```python
from datetime import date, timedelta
# set para
code = 5632
valuation_date = date(2025, 3, 3)
```

```python
# set objects
from wequant.graph_processing import PricelistFig
```

```python
Pl_fig = PricelistFig(
    code,
    start_date = valuation_date - timedelta(days=250),
    end_date = valuation_date + timedelta(days = 115)
)
#Pl_fig.add_vline_announcement_date()
Pl_fig.add_vline(valuation_date)
Pl_fig.add_trace_moving_average(5)
Pl_fig.add_trace_moving_average(25)
Pl_fig.add_trace_moving_average(75)
        

Pl_fig.fig.show()
```
