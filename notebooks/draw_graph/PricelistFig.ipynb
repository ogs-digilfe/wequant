```python
from datetime import date, timedelta
# set para
code = 9502
valuation_date = date(2026, 4, 1)
```

```python
# set objects
from wequant.graph_processing import PricelistFig
```

```python
Pl_fig = PricelistFig(
    code,
    start_date = valuation_date - timedelta(days=200),
    end_date = valuation_date + timedelta(days = 165)
)
#Pl_fig.add_vline_announcement_date()
Pl_fig.add_vline(valuation_date)
Pl_fig.add_trace_moving_average(5)
Pl_fig.add_trace_moving_average(25)
Pl_fig.add_trace_moving_average(75)
        

Pl_fig.fig.show()
```
